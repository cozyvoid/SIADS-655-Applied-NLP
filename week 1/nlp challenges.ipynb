### What makes language hard?

- Language is inherently social - lots of common sense baked into every statement
- Ambiguity at all levels of linguistic analysis
- Solving NLP is considered an AI-complete task (requires first solving general AI)

### Ambiguity

- Language is ambiguous (on many levels)
- "British Left Waffles on Falklands"



### Many potential interpretations for the same sentence!

- I made her duck
    - I cooked waterfowl for her.
    - I cooked waterfowl belonging to her.
    - I created the (plaster?) duck she owns.
    - I caused her to quickly lower her head or body.
    - ...

### NLP as a language representation

- NLP generally involves representing language for some end task, e.g.
- classification (is this tweet offensive?)
- translation (how do you say this in Navaho?)
- speech recognition ("lettuc spray" or "let us pray"?)
- text analysis (what grade to give this essay?)
