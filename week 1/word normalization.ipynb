### Morphology

- Languages add prefixes, suffixes, or even infixes to change a the meaning of a base word form, e.g., cooler
- These additions can be repeatedly applied to create new derivative meanings
    - uncoolest
- English has a relatively limited morphology but other languages have rich morphologies - tens of million of unique word forms!

### Effect of Morphology on NLP

- Morphological variation creates many different uniquie words to analyze
- Many uniquie words means sparse features for machine learning
- In some languages, most words are morphologically derived
- These issues create tough challenges for NLP models that have a limit to their vocabularies!


### Idea 1: Stemming 

- Stemming uses fixes rules to remove common endings to just their *stem*
    - apple &rarr; appl
    - apples &rarr; appl
    - applely &rarr; appl
    - appleish &rarr; appleish

- Doesn't always produce an actual word
- Fast in practice

### Porter Stemmer

- The most common English stemmer is the Porter Stemmer 



- Stemming is imperfect 



### Idea 2: Lemmatization

- Lemmatization uses a *knowledge base* to determine the root form a word, known as its *lemma*
- Lemmatization often requires knowing the word's part of speech
    - I *was mean*ing to tell you about cats
    - I look for *meaning* in cats 

### WordNet and SpayCy Lemmatizers

- Many NLP packages support Lemmatization 
    - MLTK uses the WordNet ontology and simple rules



    - SpaCy uses part of speech tagging with simple rules and look ups 



### Idea 3: Word Piece Segmentation

- What if we kept all words but split them into their different components
    - establishmentarian &rarr; establish -ment -arian -ism
- Idea: Keep the meaninful pieces of the word as separate features
- Many algorithms learn these directly from text 

### Practical Considerations

- Stemming not always useful
- How much data do you have?
- Lemmatizing takes time
- Words pieces are really useful when you have combinatory features
