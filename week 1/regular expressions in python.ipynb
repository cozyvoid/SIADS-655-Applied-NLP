### re.findall

- `findall` returns a list of all the strings that matched



### re.match 

- The `match` functon matches the whole line of text 



### The Match object 

- A call to find or match returns a `Match` object or `None` if nothing matched
- Contains useful information on what was matched
- Gives you access to any *groups* in your search

### groups

- () define a group that you can retrieve as a specific part of a larger match
- Groups are useful if you need to retrieve specific information in a larger context that has some constraints
- `match.group(0)` is always the entire match 



### Re substitutions

- This can even work with groups too
