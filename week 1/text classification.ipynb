### Text Classification

- Text classification takes some form of *text* as input and predicts some *label* for that text
    - online comment &rarr; {toxic, non-toxic}
    - news article &rarr; {real, fake-news}
- Often we use *supervised machine learning* for text classification

### Text categorization problems are everywhere



### Central questions

- What exactly is the *task* we're trying to do?
- What kind of *data* are we going to use and apply our model to?
- What are our *labels*?
- How do we *represent* text?

### Example task: sentiment analysis

- Document-level Sentiment Analysis: is the entire text *positive* or *negative* with respect to an *implicit* target?
    - Movie reviews
    - Product Reviews
    - Even some social media posts!

### There are other ways to analyze sentiment in text



- Aspect-based Sentiment Analysis: Is the sentiment for an explicit target (or aspect) positive or negative?

### Alternative Formulation: Sentiment as *tone*

- Sentiment as the positive/negative tone that a reader infers - no longer with respect to an implicit/explicit target
    - The child ran playfully through the field.
    - I'm working on my NLP homework.
    - The hard rain kept falling as we searched for our car keys in the dark.

### Building a Text Classifier

- Given training data in the form of *<x, y>* pairs, create a *function* that maps the input *x* to output *y*



- Doesn't have to be machine learning - could be as easy as using `if` and `else` statements

### The classification function

- The *classification function* that we want to learn has two different components:
    - the *representation* of the data
    - the *learning method*: How do we map the input representation to an output label?

### Representing text for classification 



- Text has to be converted to something a machine learning method can use
- One simple but powerful representation of tezt is the *bag of words*: the countsof words that it contains

### Unigrams, Bigrams, and ngrams



- The document representation doesn't just to just count words!
- *Bigrams* (two word sequences) and *trigrams* (three word) are very useful features in text classification
- *ngrams* generalize this concept to sequences of length *n*

### Weighing words by importance: TF-IDF

- Not all words are important for classification, e.g., "the" occurs in many documents
- One option is to weigh words by some function so more "important" words are scored higher
- TF-IDF is based on two ideas:
    - *Term Frequency (TF)* - words appearing more get a higher weight
    - *Inverse Document Frequency (IDF)* - words appearing in few documents are also weighted high

### Why are text classification tasks like Sentiment Analysis hard for computers?

- NLP tasks like sentiment analysis try to measure what a person is *thinking (unobserved)* from what is *written(observed)*
- In some cases, specific words can indicate the mental state that led to a sentence (*fantastic*, *must-see*, *avoid*)
- Other times, accurate classification requires *world knowledge* beyond what is evident from the text alone
