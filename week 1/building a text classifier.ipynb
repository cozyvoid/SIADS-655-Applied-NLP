### Why do we need a sparse matrix?

- A training dataset may have thousands to millions of words
- Most documents don't have all the words
- Example: 
    - 1000 documents with a vocabulary of 100000 words, and an average document length of 10
        - Dense matrix representation: 1k * 10k = 80mb
        - Sparse matrix representation: 1k * 3 * 10 = ~240kb

### Training your classifier



### Practical Considerations

- Consider what is a token for your use-case. Whitespace delimited? Only ASCII? Should emoji count? What about punctuation?

- Aggresively remove rare words (or ngrams) as features
- Lower-casing is usually helpful unless you know there are meaningfully-capitalized words "apple" vs "Apple" that matter
- Ensure you have enough data to see examples of generalizable word features 

### Practical Considerations (2)

- Easiest to start with Logistic Regression + bigrams
- Random Forests are powerful for combinatorial features
- Multilayer Perception with zero hidden layers can be used for multilabel classification (i.e., predict more than one output)
- Training on different-sized subsets of your data can help you estimate how much more data you might need
