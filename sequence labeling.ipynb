### Sequence Labeling Tasks are Everywhere in NLP 

- Part of Speech Tagging 
- Named Entity Recognition 
- Biomedical Information Extraction 
- Supersense tagging 

### What makes sequence labeling different?



### Sequence labels can be word classes 



### Sequence label can identify phrases too 

- Text sequences like named entities are commonly extracted using a BIO labeling each token as:    
    - the Beginning of a unit (e.g., a named entity)
    - being Inside of a unit 
    - being Outside of a unit 



### Sequence labeling in NLP 

- Many approaches uses *Hidden Markov Models* (HMMs)
- Maximum Entropy Markoc Models (MEMMs)
- Other more computationally-intensive appraoches include: 
    - Conditional Random Field 
    - Deep Learning 

### HMM for POS Tagging 



### Transition probabilities between parts of speech 



### Emission probabilities



### Sequence Labeling with HMMs 

- Transition and Emission probabilities are learned from text 
- Smoothing is applied to both probability tables 
- Labels are assigned using the Viterbi algorithm to decide what is the most likely sequence 
- Higher-order HMMs are possible too - but more expensive!

### Maximum Entropy Markov Model (MEMM) 



### many useful MEMM Features 

- $w_i$, $w_{i-1}$, $w_{i+1}$, $w_{i+2}$ ...
- Suffices and prefixes of $w_i$
- Presence of punctuation, capitalization 
- Topic of the current document 
- URL where the sentence is from 

### Sequence Labeling Take-aways 

- Sequence labeling allows machine models to take advantage of *correlations between the output labels* in the sequence 
- Sequence labeling tasks are frequent in NLP: POS tagging, Named Entity Recognition, Semantic Categories,...
- HMMs and MEMMs are efficient models for many NLP sequence labeling tasks
- Conditional Random Fields (CRFs) try to optimize all the output labels at once - very powerful but also slow
