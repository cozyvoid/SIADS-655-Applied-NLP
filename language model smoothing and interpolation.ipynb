# Language Model Smoothing and Interpolation

---

### Language models rely on observed text



- Probabilities are estimated based on word occurrences in a *training corpus*.
- Humans are endlessly creative so not all sequences are likely to be observed.
- The probability of an unobserved sequences is *zero*.

### Case Study: Berkeley Restaurant Project 



- An early voice assistant (1994) for helping Berkeley California residents find restaurants
- Language models are needed to resolve ambiguity in speech

### Smoothing is a family of techniques to avoid zero probabilities 



- p("I want to ear *or* get a...") = 0



- **Idea**: Redistribute the probability mass slightly so that no word has zero probability.

### Simple Approach: Additive smoothing 



### Interpolation allows us to blend two language models 



- A linear interpolation of any two language models p and q (with $\gamma \in [0,1]$) is also a valid language model.

### We can use interpolation to overcome sparsity in higher-order language models!



### Handling unknown words

- If we know all the words in advanced &rarr; Closed vocabulary task
- If new words could appear -- *Out of Vocabulary* (OOV) &rarr; Open vocabulary task 
    - These words would always have zero probability!
- Instead: create an unknown word token <UNK>
    - *Training* of <UNK> probabilities
        - Create a fixed lexicon L of size V
        - At text normalization phase, any training word not in L changed to <UNK> and we train <UNK>'s probabilities like a normal word

### Advanced UNKing 



- Words have *regular morphology* that we expect, even when we don't know what the word means
- Most systems don't use a single UNK token; UNK + *suffix* or *prefix* + UNK
