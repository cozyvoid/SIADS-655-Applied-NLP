### How do we know if two senses are distinct?

- Create a context that evokes both meanings (*zeugma*) and see if a single sense can encompass both
- Which flights *serve* breakfast?
- Does Midwest Express *serve* Philadelphia?
- (?) Does Midwest Express serve breakfast and Philadelphia?

### Zeugma isn't always a problem

- Can get productive yoking of multiple meanings 
    - [They] covered themselves with dust and glory. (Mark Twain)
    - You are free to execute your laws and your citizens as you see fit. (ST:TNG)
    - I like my *X* like I like my *Y*:*Z*


### Semantic relationships hold between senses of a word 

- Synonymy/antonymy - similarity/difference 
- Hypernynmy/hyphymy - general/more-specific kind
- Metonymy - usw of one aspect to refer to another 
- Meronymy - X is part-of Y 

### Synonymy

- Two senses of fidderent words are synonyms if their meaning is *nearly identical* 
- Words with synonymous senses can swapped without changing the meaning of a sentence
    - How *big* is the plane?
    - Am I flying on a *large* or small plane?
    - She became a *big* sister to them
    - (?) She became a *large* sister to them 

### Hyponymy 

- Sense A is a *hypo*nym of sense B if A is a subclass of B "A is a kind of B"
- *Hyper*nymy is the complementary relationship 



### Metonymy 

- Two senses describe different aspects of an entity, where teh use of the name of the entity refers to one aspect
    - The White House announced that ...
        - White House &rarr; a spokesperson for the White House 
    - The upper peninsula grows amazing apples 
        - apple (tree) &rarr; fruit of the apple tree
