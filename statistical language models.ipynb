### Language Model

- Language modeling is the task of estimating the probability of an arbitrary sequence of words *$P(w_1, ... w_n)$*
- We've seen this for one type of language model: a *unigram language model* that assumes all words are independent

- How do we add more context to estimate probability?

### Why not extend to even more context?

- Longer n-grams require even more data - huge computational requirements
- Even with more context, language generation still suffers from the challenge of *long-range dependencies*: 

> "The slide(s) that I had just uploaded to Coursera for the MLP course has (have) typos."
