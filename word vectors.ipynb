
### *Distributional* representations of words use context to encode meaning 

- Distributional methods use *vectors* that are typically learned from observing how a word is used.
- Many different approaches to encoding context.
- For all approaches, the underlying principle is that words that appear in very similar context...
    - have the *same meaning* (the *distributional hypothesis*)
    - are represented with *similar vectors*

### Defining context with co-occurrence



- Nearby words are often a good indication of meaning
```
        all you need for ___ is cards and paper
 My friends and I played ___ last night

```

- We can build a *term-term matrix* to capture these co-occurrences
    - Words are rows
    - Columns are words appearing nearby 
- Typically, we defien a *fixed-window* around an occurrence for counting

### Comparing words vectors 



- Vectors are mathematical objects, so most NLP approaches use linear algebraic methods to compare vectors
- Typically, vectors are compared using *cosine similarity*, which is the angle between two vectors 

- Other approaches like Euclidean distance are possible

### Co-occurrence-based representations

- Context provides a rich way to implicitly represent word meaning 
- The representations ultimately depend on which data you train them on and how you define context
- Many different ways to define and improve these representations
    - You already know one: *TF-IDF* for term-document matrices!
