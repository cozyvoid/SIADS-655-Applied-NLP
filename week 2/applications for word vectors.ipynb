### What can we do with vector representations of meaning?

- Many uses within NLP methods as *features* 
- *Intrinsic* applications to study language 
- *Extrinsic* applications to study culture 

### Vectors as Features for ML 

- Many applications depend on knowing the meaning of words 
    - Machine Translation
    - Parsing 
    - even Classification!
- Vectors provide useful representations of meaning
    - Similar provide useful similar vectors, which allow models to *generalize* better
    - Dense vectors are efficient representations of meaning

### Example: Classification on Vectors 



- When vectors are learned from a larger vocabulary than in training data, models can potentially generalize to examples with new words not in training if those have embeddings 

### Intrinsic Applications for Word Vectors 

- Often to used to *evaluate* whether a method has learned a good representation of word meaning 
- Tests include measuring word similarity, word analogies, or word relatedness
- Can be used to test what types of information a word vector encodes 

### Intrinsic Evaluation: Word Similarity 



- Measured by whether *vector similarities* are highly correlated with *human judgements* of semantic similarity
- Most commonly used setting is SimLex-999, with augments on a [1, 10] scale 

### Intrinsic Evaluation: Semantic Relatedness 



- Evaluated by correlation of cosine similarity with human judgements 
- Most famous dataset is WordSim-353, which has both *similarity* and *relatedness* on a [0, 10] scale
