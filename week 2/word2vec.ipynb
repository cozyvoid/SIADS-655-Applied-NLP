### word2vec 

- word2vec is a family of methods for learning dense vector representations - not "one algorithm"
- rather than count co-occurrence, *predict* which word co-occurs
- word2vec turns distributional learning into a supervised prediction task 
    - the parameters of the word2vec model become the word vectors 

### word2vec as a language model 

- Train a classifier on a binary prediction task:
    - Is word $w_1$ likely to show up near word $w_2$?
    - Looks sort of like a language model!
    - We don't actually care about the task itself
- But we'll take the parameters of the learned *classifier* as the *word embeddings*

### The word2vec suite of methods 

- Two distinct models for forming the classification task 
    - CBoW - from the *context* word, predict the *target* word
    - Skip-Gram - from the *target* word, predict the *context* word 
- Many fancier methods and hyperparameters for how to train efficiently 
    - One important idea in particular is *negative sampling* on how to create training instances

### word2vec: Skip-gram algorithm (1)







1. Treat the *target word* and *a neighboring context word* as *positive examples*. 
2. Randomly sample *other words* in the lexicon to get *negative samples*. 
3. Use feed forward neural network to train a classifier to distinguish those two cases 
4. Use the neural network weights as word vectors 

### Learning word embeddings with skip-gram

- Given a tuple (*target*, *context*)
    - e.g., (*chair*, *creak*) or (*chair*, *tire*)
- Return probability that *context* is a real context word 
    - *Intuition*: If target and content are vectors, could we estimate this by cosine similarity?
    - *Problem*: cosine similarity isn't a probability

### Summary of word2vec skip-gram learning 

- word2vec uses a shallow 1-layer neural network
- skip-gram predicts the probability of a word being observed as *context* for a *target*
- The weights of the network are updated iteratively using *stochastic gradient descent*
- after training, the weights for a word in the network become its embedding
