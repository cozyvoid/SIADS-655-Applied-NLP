### Classification based LMs allow for richer representations of context in estimating $p(w|C)$

- We can add knowledge from the whole context, document, and beyond - not just the previous word!
    - higher-order ngrams ($w_{i-1}, w{i-2}: w_{i-1}, w_{i-3}: w_{i-1}$)
    - spelling, capitalization, word eloonnnngation
    - document features (athor, topic, URL domain)
- The downside is adding many new features is potentially overfitting and generally slower training times
