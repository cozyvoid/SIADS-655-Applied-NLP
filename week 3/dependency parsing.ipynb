# Dependency Parsing 

---

### What is Dependency Parsing?

- Defined by how individual words relate to each other
    - Relationships are asymmetric
    - Relationships are types from a set of categories
- No intermediate nodes/elements 



### Which word is the *head* and which is the *dependent*?

- Many conflicting frameworks!
- Generally, the head is obligatory and dependent words are optional
- Colloquially, head words are more "important" in the sentence/phrase 



### Dependency parses form *trees* 

- A dependency structure is a directed graph G = (V, A) consisting of a set of vertices V and arc A between them. Typically contrained to form a *tree*:
    - Single root vertex with no incoming arcs 
    - Every vertex has exactly one incoming arc except root (*single head contraint*)
    - There is a uniwue path from the root to each vertex in V (*acyclic constraint*)

### Dependency Parsing Algorithms 

- Often make simplying assumptiions about hjow words relate to one another 
- Neural network-based parsers are the standard now (and fast)
- Some variations in how arcs are labeled and where they attach

### Dependencies offer a useful representation for reasoning about language 

- The binary relationships allow easy tuple extraction:
    - nsubj(Michigan, beat)
    - dobj(beat, Ohio)
- These tuples often form the core of knowledge bases and information extraction systems
