### Two main formalisms for describing and modeling syntax



### Constituency parsing 

- Words are connected through intermediate nodes that describe how *constituents* relate to one another 
- Constituency parsing often forms the base representation for semantic parsing for reasoning and answering questions 
    - Which states border states that touch a great lake?
    - How many times has Michigan outscored Ohio by more than a touchdown?
- Useful for grammer checking 

### Dependency Parsing 

- Words are directly linked in a tree based on grammatical relationships between pairs of words 
- Forms an alternative representation for reasoning about language
- Often useful in information extraction and argument mining 
- Applications in summarization 
- More support in multuiple languages
