### Dimensionality reduction on distributional representations 

- We may not need all dimensions in our representations
- Some contexts are redudant: ocean, sea (synonyms)
- Some contexts are uninformative: the, an, of 
- Working with large vectors of size |V| or |D| is expensive and slow 

### Singular Value Decomposition 

- An n x p matrix X can be decomposed into the product of three matrices 



### Singular value decomposition for dimensionality reduction 

- We can *approximate* the full matrix by only using the *k* largest singular values
- In a term-context matrix of |V| x |W|, we use the SVD to reduce word vectors to just *k* dimensions instead of |C|!
