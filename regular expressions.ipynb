### What's ELIZA doing?

- I feel Y &rarr; How often do you feel Y?
- I want Y &rarr; Suppose you got Y soon
- If Y &rarr; Do you think that it's likely that Y?
- Other tricks: 
    - Convert "my" to "your" in reply (and other pronouns)
    - Randomly produce a change of subject if no rule matche: "tell me about your mother"

### Why Regular Expressions in NLP?

- Searching for linguistic phenomena (is "consume" ever followed by "news")?
- Creating features for supervised algorithms
- Useful for morphology
- Thinking about regular expressions (as a tool) will help you think about the structure of language 

### Regular Expression: Disjunctions

- Any string is a regular expression that's an exact match, e.g., "woodchuck"
- What if we wanted to find all of these words?

$$$
Woodchuck
woodchuck
$$$

- The *disjuction operator* [] lets us specify *multiple* matching characters:

$$$
"[wW]oodchuck"
$$$

- We can specify whole *ranges* of matches using the '-' character:
"[a-z]" matches a, b, c, ..., z.

### Regular Expressions: Negation in Disjunctions

- Sometimes we want to *not* match. We use the '^' at the start of a [] expression to anti-match

### Regular Expressions: Groups

- *Groups* allow us to reference parts of matching
- We create groups by putting () around a subexpression, e.g., "My name is ([a-zA-Z]+)"
- We can reference thme for string replacement, which is "\1" for the first group, "\2" for the second, and so on
    - Group references can vary by programming language so check
