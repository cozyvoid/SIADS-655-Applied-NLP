# Syntactic Libraries 

----

### Natural Language Toolkit (NLTK)

- https://www.nltk.org/
- General purpose NLP library
- Implementations for many syntactic and semantic algorithms
    - POS Tagging, All forms of Parsing, Coreference, NER, Word Sense Disambiguation,...
- Not many deep learning or word vector models 
- Built in datasets and corpora 
- Bindings to some other libraries like CoreNLP
- Mostly pure Python and designed for education use (tutorials!)
    - Often behind state of the art

### SpaCy

- https://spacy.io/
- Implements many NLP methods for syntactic and some semantic analyses
    - POS Tagging, Parsing, Coreference, NER, ...
- Python implementation backed by C (very fast)
- Multiple models for different domains ("web", "news")
- Multilingual support

### CoreNLP

- https://stanforddnlp.github.io/CoreNLP/

- State-of-the-art models for many syntanctic and discourse tasks
    - POS Tagging, Parsing, Coreference, NER, Quote Recognition, Entity Linking, Information Extraction
- Pluggable pipeline
- Written in Java with bindings for Python
- Regularly updated with each new NLP paper 
- Multilingual support 

### Stanza 

- https://stanfordnlp.github.io/stanza/
- State-of-the-art models for many syntactic and discourse tasks 
    - POS Tagging, Parsing, Coreference, NER, Quote Recognition, Entity Linking, Information Extraction 
- Written in python using deep learning (GPU is a huge boost)
- Bindings to CoreNLP
- More likely to close to state of the art at any given moment

### TextaCy 

- https://github.com/chartbeat-labs/textacy 
- Useful lNLP and Information Extraction methods on top of SpaCy
    - e.g., noun checking, POS Regex, extracting dependency triples
- Provides programmatic access to common corpora (like NLTK)
- Other useful tooks like estimating a text's "grade level" or constructing word-document matrices
