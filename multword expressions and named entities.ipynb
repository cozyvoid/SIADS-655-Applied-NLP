# Multiword Expressions and Named Entities 

---

### Language isn't just made of words 



### Named Entities and MWEs  

- Named Entities are things that can be referred to with a proper name 
    - People, famous animals 
    - Places, famous buildings
    - Businesses, Organizations 
    - Movies, Books, Paintings 
- A multiword expression (MWE) is a sequence of words whose meaning isn't fully predictable from its component words 

### Many types of Multiword Expressions 

- Noun compound 
    - support vector machine, machine learning
- Idioms 
    - spill the beans, get into the weeds, black and white 
- Verbal constructions 
    - take a shower, go out to eat 
- Fixed phrases 
    - inter alia, ad hoc, ex ante 

### Recognizing named entities and MWEs 

- Many approaches use a list of known named entities and MWEs 
    - often called a *gazetteer* 
- Recognizing MWEs is harder due to many new MWEs 
    - Build a list of MWEs by findings frequent Noun-Noun compounds 
    - other statistical approaches for finding noun-pairs 
- Several approaches use advanced *sequence labeing* approaches to identify named entities 

### Recognizing MWEs pairs using PMI 

- Point-wise mutual information can identify word pairs that co-occur more frequently than expected by chance 

- Other approaches use statistical test like $x^2$ or a t-test 
- Harder to extend to longer phrases 

### Practical Implications 


- Recognizing named entities is critical for many downstream applications like formation extraction, summarization, and machine translation 
- Most NLP approaches focus on finding Named Entities, but MWEs can be important, especially in domains with lots of jargon 
- Running Named Entity Recognition or merging MWEs can be a useful preprocessing step for learning vector representations
