### Event-based Semantic Roles



- Verbs like "write", "buy", or "tell" form *predicates* for events
- Events can have *arguments* that specify their *role* in the event
    - Who is the *cause* of the event? What is the *result*? *How* was it done?

### Many definitions of roles!



- Verb-specific role: "Author"
- A more abstract role: the volitional cause of the event
- Very abstract role: the "do-er" (vs. "the receiver of the doing")

### Thematic roles capture common roles across different verbs

The *student* wrote the *program* in python using PyCharm.

The *musician* played a *sonata* on the guitar.

The *lawyer* delivered a closing *argument*. 

- student, musician, and lawyer all have agency, are acting of their own volition, and directly cause the events to transpire
- program, sonata, and argument all are things that are the results of their events

### Roles have many uses in NLP

- Semantic role-labled data is an intermediate representation of the semantics of a sentence
- Helpful for Question Answering, Machine Translation, Information Extraction

### Challenges in Defining Roles 

- What are the boundaries and criteria?
    - Can a vacuum have volition?
- Some definitions have opted for very coarse roles: Proto-roles (Dowty, 1991)
    - Proto-Agent vs. Proto-Patient
- Some definitions have many more roles, e.g., FrameNet (Baker et al., 1998)
