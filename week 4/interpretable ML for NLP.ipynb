# Interpretable ML for NLP 

---

### How do we know our classifier is making a good prediction? 

- Initial Answer: Look at what features it cares about!
- For some classifiers like Logistic Regression, this approach is feasible: $/beta$ vetor shows you which features the model is weighting
- What about for a Random Forest? Deep Learning?
- *Explainable ML* aims to generate human-readable explanations for why a model made its decision 

### Key Ideas for MLK Explanations 

- Practioners can understand the explanation!
- Faithful to what the model actually does 
- Ideally, applicable across multiple models 
- Caveat: Lots of discussion for what makes a "good" explanation
    - Who are explanations useful for and what purpose do they serve?

### One-approach: Local Interpretable Model-Agnostic Explanations (LIME)



- Start with a given classifier who has an unknown decision boundary (sarcastic, non-sarcastic)
- Pick some particular instance to explain 
- Locally approximate the decision boundary of the global model 
    - Train a simple model that is globally bad by locally faithful 
    - Use the simple model to explain decision 

### Explainable ML is a hot topic 

- Many new approaches each year, often with user-friendly libraries 
    - LIME 
    - SHAP 
    - ELI5 
- As a practioner, use these techniques to build trust in your complex NLP models 
    - Are my features reasonable? Is the model looking at the right things?
