# WordNet 

---

### Dictionaries as knowledge bases 



### WordNet 

- Lexical databases for *nouns*, *verbs* and *adjectives/adverbs*.
- Each word sense is arranged in a *synset* (category of near-synonyms) and each synset is related to others in terms of their sense relations. 

### One lemma can be in multiple synsets 



### Each sense has his own hypernyms 



### Noun Relations in WordNet 



### Verb senses too have relationships 



### WordNet Structure Encodes Meaning 

- similarity("tip", "wage") vs. similarity("tip", "peak")
- Distributional approaches like word2vec represent the meaning of "tip" in a single vector 



### WordNet can be used for semantic similarity beyond just words 



### WordNet in the Wild: SemCor

- SemCor: 352 texts from the Brown Corpus where all contents words are labeled with WordNet 1.6 senses 
    - Brown Corpus is American English from the early 1960's
    - ~200k content words in total 
    - Later versions with updates annotations (WordNet 3.0)



### WordNet is available in may languages

- 50+ WordNets in languages like Arabic, Bantu, Chinese, French, Irish, Marathi, Sanskrit, and Tamil 
- All use similar structure 
- Not all link across languages
