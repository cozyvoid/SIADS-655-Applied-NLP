##### Frame semantics aim to capture this type of background information that readers naturally infer.

### Semantic frames 

- A frame is like a *knowledge structure* that reflects stereotypical information about an event
    - Who is involved? What happens? When does it happen?...
- Designed to help NLP systems brdige the gaps from text to the "real world"
    - Frames can be cued by even a single word!

### Propnet compared to FrameNet 

- PropBank reflects the structure and types of arguments for an individual *verb sense*
    - increase.v:
        - Arg0: CAuse of the increase
        - Arg1: Thing that is increased 
        - Arg2: Amount of increase
        - ...

- FrameNet reflects the structure and types of arguments for *frames*, which are signaled by a *lexical unit*
    - `cooking_event`
        - `cook`
        - `produced_food`
        - `heating_instrument`
        - ...

### Semantic Frame: `cooking_event`

- Lexical units: words that signal the frame's presence 

bake.v, baking.v, concoct.v, cook up.v, cook.n, cook.v, cooking.n, fix.v, fry.v, grill.v, grilling.n, make.v, preparation.n, prepare.v, put together.v, roast.v, roasting.n, whip up.v 

These are the core elements of the frame. 

- Core Frame Elements: 
    - Cook: the Cook prepares the `Produced_food`
    - Produced_food: the `Produced_food` is the result of a Cook's efforts
    - Container: the Container that holds the food being produced
    - Degree: the Degree to which an event occurs 
    - Heating instrument: the entity that directly supplies heat to the Food

### Lexical units aren't just verbs 

- For the "change in position on a scale" frame, example units...
    - verbs: increase, advance, jump, slide 
    - nouns: escalation, incrase, growth, shift 
    - adjectives: accelerated, growing
    - adverbs: increasingly
