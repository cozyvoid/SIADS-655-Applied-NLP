### What exactly is a part of speech? 

- A part of speech is a *category* where its words...
    - appear in similar syntactic contexts 
    - perform similar types of functions in sentences
    - have similar morphological transformations 
- A part of speech is a *distribution* over these properties!

### Languages can differ in their parts of speech 

- Some languages mostly don't differentiate between noun/adjective (Quechua) or noun/verb (Tongan)
- Some languages have very few adjectives (Thai)
- Some classes are largely closed in a language - e.g., verbs and adjectives are closed in Japanese 
- These claims depends on how you define a part of speech and are an open area of Linguistics!
- Some work in NLP on creating a universal POS tag set (Petrov, Das, McDonald; LREC 2012)
