### Word Sense Disambiguation 

- Word Sense Disambiguation aims to identify which sense of a word is present is a particular usage

"Historian_1 have come_4 to recognize_4 two_1 cardinal_1 facts_1 concerning_2 nationalism_1 and international_2 influence_3"

- Two appraoches to WSD:
    - All Words Models label all content words jointly with their senses in a fixed sense inventory (e.g., WordNet)
    - Single Word Models label a specific content word (e.g., "bank")

### Word Sense Disambiguation (WSD)

- WSD is a *difficult* task, so many approaches
- knowledge-based methods rely on information from some external source, e.g., dictionaries or semantic networks
- Supervised methods use traditional machine learning 
- Semi-supervised methods bootstrap from a small amount of labeled data


### Knowledge-based methods

- Many sense inventories like WordNet encode rich information that can be leveraged for WSD
    - Definitions
    - Glosses (example sentences)
    - Semantic networds of how senses relate
- Knowledge-based methods predict the sense by comparing information in the knowledge base

### Supervised WSD 

- Treat WSD as a supervised learning problems
    - Single-word methods: Regular Classifiers, HMM, MEMM
    - All-words Methods: CRF, Deep Learning 
- All rely on training data, which is in short supply
    - SemCor 200k tokens - not all senses covered 
    - Multiple Approaches to prodedurally generate training data 

### Supervised WSD Features 

- Words in specific positions before or after the target, e.g., "river *bank*"
- Words in the context
- Topical information
- Syntactic modifiers
- Language model predictions for the target's context
- Distributional vectors of words, context, document...

### Semi-supervised WSD 

1. Label an initial set of seed examples
    1. Sentences from a knowledge base
    2. Sentences containing a sense-specific n-gram ("river bank")
2. Using a large set of *unlabeled contexts*, repeat until convergence:
    1. Train supervised classifier on current label contexts
    2. Classify all examples and add *high-confidence predictions* to the set of labeled contexts

### Simplifying Heuristic: One sense per discourse

- If a word appears *multiple times* in a document, it's usually with the *same sense* (Gale et al. 1992)
    - Articles about restaurant tips typically don't talk about mountain peaks or hints
- Label all instances with a classifier, take the majoritty label or the label with the highest confidence
