# Evaluating Language Models 

---

### How do we know if we've learned a good model of language?

- Evaluate using a standard machine learning setup: <train, development, test datasets>
- Intrinsic tests of language models 
- Extrinsic tests of language models for downstream tasks 

### Fitting your language model 



- The language model is fit using the training data 
- The development data is used to pick between models with different hyperparameters, e.g., $|V|, \gamma_1, \gamma_2$, max n-gram
- The dev and test data may contain unseen words, so these test your model's ability to generalize to new sequences.

### Intrinsic Test: Perplexity

- How surprised is the model by *unseen sentences*?
- Perplexity is the inverse probability of the test data, normalized by its length N: 



- Perplexity can only be compared between models on the same dataset.
- Typically only used for model selection - not a highly meaningful metric by itself.

### Extrinsic Evaluations



- A better comparison of two language models is whether they benefit *downstream task*.
- Example: spell checking 
    - Which of two models results in more corrections?
- Other possibilities
    - Machine Translation
    - Search Query Suggestions
