# Semantic Role Labeling

---

### How to label roles?



- treat it as a supervised learning task!
- Exampple using data from PropBank

### PropBank

- Extra annotation of the Penn TreeBank, which waws originally designed for parsing
- Each verb has its own roles, but these loosely align: 
    - Arg0: Proto-Agent: volitional, cause, movers, sentience...
    - Arg1: Proto-Patient: things that is affected, changes, stationary
    - Arg2: benefactive, instrument, attribute, or end state (mostly)
    - Arg3: statrting point, benefactice, instrument, or attribute (mostly)
    - Arg4: ...

### PropBank-labeled sentence for *increase*

- Arg0: Cause of the increase (proto-agent!)
- Arg1: Thing that is increased (proto-patient!)
- Arg2: Amount of increase
- Arg3: Starting point prior to increase
- Arg4: Ending point after increase



### Labeling sentences with frames



- General algorithm: 
    - Constituency parse the sentence
    - For each *predicate* (recursively)
        - *Featurize* the sentence relative to the predicate
        - *Predict* the role (if any) of the predicate 

### Semantic Role Labeling is Challenging!

- Too many predicates to classify all 0 - pruning is needed!
- Many types of arguments makes leveraging cross-grame labels difficult
    - Simpler schemes like Proto-Roles have a huge advantage here
- Training data is growing but still limited
     - PropBank annotations are available in many languages!
- Neural approaches are growing in usage - but still data-limited!
