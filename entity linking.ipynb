### Entity Linking

- Given some mention(s) of an entity in text, identify the canonical reference for that entity in an ontology.
    - Wikipedia is often the ontology
    - Other ontologies useful for entity linking in a domain (medical, pharmacological, rock stars)
- Similar to Word Sense Disambiguation as task

### Entity Linking as a task

- Often broken into two phases:
    - *Mention detection* - which elements in a text refer to this entity and what other names might this referent go by?
        - Coreference resolution commonly used 
- *Mention disambiguation* - given all the mentions, who/what is the entity
    - Dependent on underlying knowledge base.
    - Can use many of tghe approaches we described for WSD.

### Entity Linking in Practice

- Entity Linking (EL) is useful when needing to reason about texts
    - Question answering: "when did Michael Jordan get his PhD?"
- Most NLP libraries don't have EL built in 
- Dedicated EL libraries exist for specific ontologies, e.g., Wikiepedia, BabelNet
